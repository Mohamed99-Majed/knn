```python
import pandas as pd
import numpy as np
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import KFold
from sklearn.preprocessing import MinMaxScaler 
from sklearn.pipeline import Pipeline
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('hotel_bookings_cleaned.csv')
df
```

```python
print(df.columns.tolist())
```

```python
X = df.drop(columns=['is_canceled', 'reservation_status', 'reservation_status_date'])
y=df['is_canceled']            
```

```python
kf=KFold(n_splits=5,shuffle=True,random_state=42)
```

```python
texts_cols = X.select_dtypes(include=['object', 'category']).columns
nums_cols = X.select_dtypes(include=['int64', 'float64']).columns

prep = ColumnTransformer(
    transformers=[
        ('num', MinMaxScaler(), nums_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), texts_cols)
   
    ]
)
```

```python
pipeline = Pipeline(steps=[
    ('preprocessor', prep),
    ('model', KNeighborsClassifier())
])
```

```python
param_grid = {
    'model__n_neighbors': [3, 5, 7, 9],
    'model__metric': ['euclidean', 'manhattan']
}

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
   
)
```

```python
grid_search.fit(X, y)

print("Best Parameters:", grid_search.best_params_)
print("Best Accuracy Score:", grid_search.best_score_ * 100)
```
